```python
from scipy.stats import chi2_contingency

# Your A/B test counts
control_users = 60148
control_conversions = 38516

variant_a_users = 19927
variant_a_conversions = 12702

variant_b_users = 19925
variant_b_conversions = 12817
```

```python
contingency_a = [
    [control_conversions, control_users - control_conversions],
    [variant_a_conversions, variant_a_users - variant_a_conversions]
]

chi2_a, p_a, dof_a, expected_a = chi2_contingency(contingency_a)

print("Control vs Variant_A")
print("Chi-square:", round(chi2_a, 4))
print("p-value:", p_a)
```

```python
contingency_b = [
    [control_conversions, control_users - control_conversions],
    [variant_b_conversions, variant_b_users - variant_b_conversions]
]

chi2_b, p_b, dof_b, expected_b = chi2_contingency(contingency_b)

print("Control vs Variant_B")
print("Chi-square:", round(chi2_b, 4))
print("p-value:", p_b)
```

```python
alpha = 0.05

if p_a < alpha:
    print("Variant_A difference is statistically significant.")
else:
    print("Variant_A difference is NOT statistically significant.")

if p_b < alpha:
    print("Variant_B difference is statistically significant.")
else:
    print("Variant_B difference is NOT statistically significant.")
```
